```python
import pandas as pd

PLAY_STORE_FILE = "play Store Data.csv.xlsx"

apps = pd.read_excel(PLAY_STORE_FILE)

print("Dataset loaded successfully!")
print("Shape:", apps.shape)
print("\nColumns:")
print(apps.columns.tolist())
```

```python
# Check dataset structure
print("Number of rows:", len(apps))
print("Number of columns:", len(apps.columns))
print("\nColumn names:")
for i, col in enumerate(apps.columns, 1):
    print(i, ":", col)
```

```python
# Check missing values and data types

print("DATA TYPES:")
print(apps.dtypes)

print("\nMISSING VALUES:")
print(apps.isnull().sum())
```

```python
# Check for duplicate records

duplicate_count = apps.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Remove duplicate rows

before = len(apps)

apps = apps.drop_duplicates().copy()

after = len(apps)

print("Rows before removing duplicates:", before)
print("Duplicate rows removed:", before - after)
print("Rows after removing duplicates:", after)
```

```python
# Clean and convert numeric columns

# Reviews
apps["Reviews"] = pd.to_numeric(apps["Reviews"], errors="coerce")

# Installs: remove '+' and ',' and convert to numeric
apps["Installs"] = (
    apps["Installs"]
    .astype(str)
    .str.replace("+", "", regex=False)
    .str.replace(",", "", regex=False)
)
apps["Installs"] = pd.to_numeric(apps["Installs"], errors="coerce")

# Size: convert MB/GB values to MB
def convert_size(value):
    if pd.isna(value):
        return None

    value = str(value).strip()

    if value.endswith("M"):
        return float(value[:-1])

    if value.endswith("k"):
        return float(value[:-1]) / 1024

    if value.endswith("G"):
        return float(value[:-1]) * 1024

    return None

apps["Size_MB"] = apps["Size"].apply(convert_size)

# Price: remove '$' and convert to numeric
apps["Price"] = (
    apps["Price"]
    .astype(str)
    .str.replace("$", "", regex=False)
)
apps["Price"] = pd.to_numeric(apps["Price"], errors="coerce")

print("Numeric columns cleaned successfully.")
print("\nData types after cleaning:")
print(apps[["Reviews", "Installs", "Size_MB", "Price"]].dtypes)
```

```python
# Check missing values after numeric conversion

print("Missing values after cleaning:")
print(apps[["Reviews", "Installs", "Size_MB", "Price"]].isnull().sum())
```

```python
# Inspect rows with missing numeric values

print("Rows with missing Reviews:")
print(apps[apps["Reviews"].isna()][["App", "Category", "Reviews", "Installs", "Size", "Price"]])

print("\nRows with missing Installs:")
print(apps[apps["Installs"].isna()][["App", "Category", "Reviews", "Installs", "Size", "Price"]])

print("\nRows with missing Price:")
print(apps[apps["Price"].isna()][["App", "Category", "Reviews", "Installs", "Size", "Price"]])
```

```python
# Remove the row with missing critical numeric values

before = len(apps)

apps = apps.dropna(subset=["Reviews", "Installs", "Price"]).copy()

after = len(apps)

print("Missing-value rows removed:", before - after)
print("Rows before:", before)
print("Rows after:", after)

print("\nRemaining missing values:")
print(apps[["Reviews", "Installs", "Size_MB", "Price"]].isna().sum())
```

```python
# Clean text columns

text_columns = ["App", "Category", "Type", "Content Rating", "Genres"]

for col in text_columns:
    apps[col] = apps[col].astype(str).str.strip()

print("Text columns cleaned successfully.")

print("\nMissing values in important columns:")
print(apps[text_columns].isna().sum())
```

```python
# Clean Last Updated and Android Version columns

# Convert Last Updated to datetime
apps["Last Updated"] = pd.to_datetime(
    apps["Last Updated"],
    errors="coerce"
)

# Clean Android Version
apps["Android Ver"] = (
    apps["Android Ver"]
    .astype(str)
    .str.extract(r"(\d+(?:\.\d+)?)")[0]
)

apps["Android Ver"] = pd.to_numeric(
    apps["Android Ver"],
    errors="coerce"
)

print("Date and Android Version columns cleaned successfully.")

print("\nData types:")
print(apps[["Last Updated", "Android Ver"]].dtypes)

print("\nMissing values:")
print(apps[["Last Updated", "Android Ver"]].isna().sum())
```

```python
# Restore actual missing values that became the text "nan"

text_columns = ["App", "Category", "Type", "Content Rating", "Genres"]

for col in text_columns:
    apps[col] = apps[col].replace("nan", pd.NA)

print("Missing values after correction:")
print(apps[text_columns].isna().sum())
```

```python
# Remove rows missing essential categorical information

before = len(apps)

apps = apps.dropna(
    subset=["App", "Category", "Type", "Content Rating"]
).copy()

after = len(apps)

print("Rows removed:", before - after)
print("Rows remaining:", after)

print("\nRemaining missing values:")
print(apps[["App", "Category", "Type", "Content Rating", "Genres"]].isna().sum())
```

```python
# Final dataset validation

print("FINAL DATASET SUMMARY")
print("=" * 40)

print("Rows:", apps.shape[0])
print("Columns:", apps.shape[1])

print("\nData types:")
print(apps.dtypes)

print("\nMissing values:")
print(apps.isna().sum())

print("\nDuplicate rows:")
print(apps.duplicated().sum())
```

```python
# ============================================================
# TASK 1 — BUBBLE CHART
# Load User Reviews dataset
# ============================================================

USER_REVIEWS_FILE = "User Reviews.csv"

reviews = pd.read_csv(USER_REVIEWS_FILE)

print("User Reviews Dataset loaded successfully!")
print("Shape:", reviews.shape)

print("\nColumns:")
print(reviews.columns.tolist())

print("\nFirst 5 rows:")
display(reviews.head())
```

```python
# ============================================================
# TASK 1 — PREPARE SENTIMENT SUBJECTIVITY
# Calculate average subjectivity for each app
# ============================================================

app_subjectivity = (
    reviews
    .dropna(subset=["Sentiment_Subjectivity"])
    .groupby("App", as_index=False)["Sentiment_Subjectivity"]
    .mean()
    .rename(columns={
        "Sentiment_Subjectivity": "Avg_Subjectivity"
    })
)

print("Average subjectivity calculated successfully.")

print("Number of apps with subjectivity data:", len(app_subjectivity))

print("\nFirst 5 results:")
display(app_subjectivity.head())
```

```python
# ============================================================
# TASK 1 — MERGE SUBJECTIVITY WITH PLAY STORE DATA
# ============================================================

task1_data = apps.merge(
    app_subjectivity,
    on="App",
    how="left"
)

print("Task 1 datasets merged successfully!")

print("Rows:", task1_data.shape[0])
print("Columns:", task1_data.shape[1])

print("\nSubjectivity missing values:")
print(task1_data["Avg_Subjectivity"].isna().sum())

print("\nSample of merged data:")
display(
    task1_data[
        ["App", "Category", "Rating", "Reviews",
         "Size_MB", "Installs", "Avg_Subjectivity"]
    ].head()
)
```

```python
# ============================================================
# TASK 1 — CHECK APP NAME MATCHING
# ============================================================

play_store_apps = set(apps["App"].dropna().unique())
review_apps = set(reviews["App"].dropna().unique())

matched_apps = play_store_apps.intersection(review_apps)

print("Unique Play Store apps:", len(play_store_apps))
print("Unique Review apps:", len(review_apps))
print("Matching app names:", len(matched_apps))

print("\nPlay Store apps without review data:",
      len(play_store_apps - review_apps))

print("Review apps not in Play Store:",
      len(review_apps - play_store_apps))
```

```python
# ============================================================
# TASK 1 — APPLY ALL REQUIRED FILTERS
# ============================================================

allowed_categories = [
    "GAME",
    "BEAUTY",
    "BUSINESS",
    "COMICS",
    "COMMUNICATION",
    "DATING",
    "ENTERTAINMENT",
    "SOCIAL",
    "EVENTS"
]

task1_data = task1_data[
    # Rating > 3.5
    (task1_data["Rating"] > 3.5)

    # Required categories
    & (task1_data["Category"].isin(allowed_categories))

    # Reviews > 500
    & (task1_data["Reviews"] > 500)

    # Installs > 50K
    & (task1_data["Installs"] > 50000)

    # App name must not start with X, Y or Z
    & (~task1_data["App"].str.lower().str.startswith(
        ("x", "y", "z"), na=False
    ))

    # App name must not contain the letter S
    & (~task1_data["App"].str.lower().str.contains(
        "s", na=False
    ))

    # Sentiment subjectivity > 0.5
    & (task1_data["Avg_Subjectivity"] > 0.5)

    # Size must be available for the X-axis
    & (task1_data["Size_MB"].notna())
].copy()

print("Task 1 filtered rows:", len(task1_data))

print("\nCategories remaining:")
print(task1_data["Category"].value_counts())

print("\nMissing values in required chart columns:")
print(
    task1_data[
        ["App", "Category", "Size_MB", "Rating",
         "Installs", "Reviews", "Avg_Subjectivity"]
    ].isna().sum()
)
```

```python
# ============================================================
# TASK 2 - INTERACTIVE PLOTLY CHOROPLETH MAP
# Global App Installs by Category
# Display only between 6 PM and 8 PM IST
# ============================================================

import pandas as pd
import plotly.express as px
from datetime import datetime
from zoneinfo import ZoneInfo
import os

# ------------------------------------------------------------
# 1. Check the current dataset
# ------------------------------------------------------------

print("Task 2 - Dataset check")
print("=" * 50)

print("Rows:", len(apps))
print("Columns:")
print(apps.columns.tolist())

# ------------------------------------------------------------
# 2. Check for country/location information
# ------------------------------------------------------------

location_columns = [
    col for col in apps.columns
    if any(word in col.lower()
           for word in ["country", "region", "location", "continent"])
]

print("\nCountry/Region/Location columns found:")
print(location_columns)

# ------------------------------------------------------------
# 3. Check for GeoJSON files
# ------------------------------------------------------------

geojson_files = [
    file for file in os.listdir()
    if file.lower().endswith(".geojson")
    or file.lower().endswith(".json")
]

print("\nGeoJSON/JSON files found:")
print(geojson_files)

# ------------------------------------------------------------
# 4. Apply Task 2 category requirements
# ------------------------------------------------------------

task2_data = apps.copy()

# Remove missing values required for analysis
task2_data = task2_data.dropna(
    subset=["App", "Category", "Installs"]
).copy()

# ------------------------------------------------------------
# 5. Remove categories beginning with A, C, G or S
# ------------------------------------------------------------

task2_data = task2_data[
    ~task2_data["Category"]
    .astype(str)
    .str.upper()
    .str.startswith(("A", "C", "G", "S"))
].copy()

# ------------------------------------------------------------
# 6. Keep apps with more than 1 million installs
# ------------------------------------------------------------

task2_data = task2_data[
    task2_data["Installs"] > 1_000_000
].copy()

print("\nRows after Task 2 filters:", len(task2_data))

# ------------------------------------------------------------
# 7. Find the TOP 5 APP CATEGORIES
# ------------------------------------------------------------

top_categories = (
    task2_data.groupby("Category")["Installs"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print("\nTop 5 categories:")
print(top_categories)

# Keep only top 5 categories
task2_data = task2_data[
    task2_data["Category"].isin(top_categories.index)
].copy()

print("\nCategories remaining:")
print(task2_data["Category"].value_counts())

# ------------------------------------------------------------
# 8. Prepare category-level install data
# ------------------------------------------------------------

category_installs = (
    task2_data.groupby("Category", as_index=False)["Installs"]
    .sum()
    .sort_values("Installs", ascending=False)
)

print("\nCategory-level install totals:")
print(category_installs)

# ------------------------------------------------------------
# 9. IMPORTANT ANOMALY CHECK
# ------------------------------------------------------------

if len(location_columns) == 0:
    
    print("\n" + "=" * 60)
    print("ANOMALY DETECTED")
    print("=" * 60)
    
    print(
        "\nThe Google Play Store dataset does not contain "
        "a country, region, or location column."
    )
    
    print(
        "\nA GeoJSON file is also not available in the current "
        "Jupyter folder."
    )
    
    print(
        "\nTherefore, country-level Choropleth mapping cannot "
        "be created from the supplied dataset without inventing "
        "geographic information."
    )
    
    print(
        "\nAccording to the guidance received from ElevanceSkills, "
        "this anomaly should be mentioned in the internship report."
    )
    
    print("\nTask 2 data preparation completed successfully.")
    print("Country-level mapping requires geographic data.")

else:
    
    print("\nLocation information detected.")
    
    # --------------------------------------------------------
    # 10. Prepare country-level data
    # --------------------------------------------------------
    
    location_col = location_columns[0]
    
    print("Using location column:", location_col)
    
    country_data = (
        task2_data.groupby(
            [location_col, "Category"],
            as_index=False
        )["Installs"]
        .sum()
    )
    
    # Keep only records above 1 million installs
    country_data = country_data[
        country_data["Installs"] > 1_000_000
    ].copy()
    
    print("\nCountry-level data prepared:")
    print(country_data.head())

    # --------------------------------------------------------
    # 11. Check GeoJSON
    # --------------------------------------------------------
    
    if len(geojson_files) == 0:
        
        print("\nANOMALY:")
        print(
            "Location data exists, but no GeoJSON file was "
            "found in the current folder."
        )
        
        print(
            "\nPlease mention this limitation in the internship report."
        )
        
    else:
        
        geojson_file = geojson_files[0]
        
        print("\nUsing GeoJSON file:", geojson_file)
        
        # ----------------------------------------------------
        # 12. Create interactive Choropleth
        # ----------------------------------------------------
        
        fig = px.choropleth(
            country_data,
            locations=location_col,
            locationmode="country names",
            color="Installs",
            hover_name=location_col,
            hover_data=["Category", "Installs"],
            color_continuous_scale="Viridis",
            title="Global App Installs by Category"
        )
        
        fig.update_layout(
            title_x=0.5,
            template="plotly_white"
        )
        
        # ----------------------------------------------------
        # 13. Display only between 6 PM and 8 PM IST
        # ----------------------------------------------------
        
        current_time = datetime.now(
            ZoneInfo("Asia/Kolkata")
        ).time()
        
        start_time = current_time.replace(
            hour=18,
            minute=0,
            second=0,
            microsecond=0
        )
        
        end_time = current_time.replace(
            hour=20,
            minute=0,
            second=0,
            microsecond=0
        )
        
        if start_time <= current_time < end_time:
            
            fig.show()
            
        else:
            
            print(
                "\nTask 2 Choropleth Map is available only "
                "between 6:00 PM and 8:00 PM IST."
            )
```

```python
# ============================================================
# TASK 3 - TIME SERIES LINE CHART
# Total Installs Over Time by App Category
# Available only between 6 PM and 9 PM IST
# ============================================================

import pandas as pd
import plotly.graph_objects as go
from datetime import datetime
from zoneinfo import ZoneInfo

# ------------------------------------------------------------
# 1. Make a copy of the cleaned dataset
# ------------------------------------------------------------

task3_data = apps.copy()

# ------------------------------------------------------------
# 2. Make sure Last Updated is datetime
# ------------------------------------------------------------

task3_data["Last Updated"] = pd.to_datetime(
    task3_data["Last Updated"],
    errors="coerce"
)

# ------------------------------------------------------------
# 3. Remove rows with missing required values
# ------------------------------------------------------------

task3_data = task3_data.dropna(
    subset=["App", "Category", "Reviews", "Installs", "Last Updated"]
).copy()

# ------------------------------------------------------------
# 4. Apply Task 3 filters
# ------------------------------------------------------------

# Reviews greater than 500
task3_data = task3_data[
    task3_data["Reviews"] > 500
].copy()

# App name must NOT start with X, Y or Z
task3_data = task3_data[
    ~task3_data["App"]
    .astype(str)
    .str.upper()
    .str.startswith(("X", "Y", "Z"))
].copy()

# App name must NOT contain the letter S
task3_data = task3_data[
    ~task3_data["App"]
    .astype(str)
    .str.contains("S", case=False, na=False)
].copy()

# App category must start with E, C or B
task3_data = task3_data[
    task3_data["Category"]
    .astype(str)
    .str.upper()
    .str.startswith(("E", "C", "B"))
].copy()

# ------------------------------------------------------------
# 5. Create monthly period
# ------------------------------------------------------------

task3_data["Month"] = (
    task3_data["Last Updated"]
    .dt.to_period("M")
    .dt.to_timestamp()
)

# ------------------------------------------------------------
# 6. Calculate total installs by month and category
# ------------------------------------------------------------

monthly_data = (
    task3_data
    .groupby(["Month", "Category"], as_index=False)["Installs"]
    .sum()
    .sort_values(["Category", "Month"])
)

# ------------------------------------------------------------
# 7. Calculate month-over-month growth
# ------------------------------------------------------------

monthly_data["MoM_Growth"] = (
    monthly_data
    .groupby("Category")["Installs"]
    .pct_change() * 100
)

# ------------------------------------------------------------
# 8. Translate categories for graph display
# ------------------------------------------------------------

category_translation = {
    "BEAUTY": "सौंदर्य",          # Hindi
    "BUSINESS": "வணிகம்",         # Tamil
    "DATING": "Partnersuche"      # German
}

monthly_data["Display_Category"] = (
    monthly_data["Category"]
    .map(category_translation)
    .fillna(monthly_data["Category"])
)

# ------------------------------------------------------------
# 9. Display filtered information
# ------------------------------------------------------------

print("Task 3 data prepared successfully!")
print("=" * 55)

print("\nRows after all Task 3 filters:")
print(len(task3_data))

print("\nCategories remaining:")
print(task3_data["Category"].value_counts())

print("\nMonthly data sample:")
print(monthly_data.head(10))

# ------------------------------------------------------------
# 10. Create Plotly time-series chart
# ------------------------------------------------------------

fig = go.Figure()

categories = monthly_data["Category"].unique()

for category in categories:

    category_data = monthly_data[
        monthly_data["Category"] == category
    ].sort_values("Month").copy()

    display_name = category_data["Display_Category"].iloc[0]

    # --------------------------------------------------------
    # Line chart
    # --------------------------------------------------------

    fig.add_trace(
        go.Scatter(
            x=category_data["Month"],
            y=category_data["Installs"],
            mode="lines+markers",
            name=display_name,
            hovertemplate=(
                "<b>Category:</b> " + display_name +
                "<br><b>Month:</b> %{x|%B %Y}" +
                "<br><b>Total Installs:</b> %{y:,.0f}" +
                "<br><b>MoM Growth:</b> %{customdata:.2f}%"
                "<extra></extra>"
            ),
            customdata=category_data["MoM_Growth"]
        )
    )

    # --------------------------------------------------------
    # Shade areas where MoM growth exceeds 20%
    # --------------------------------------------------------

    for i in range(1, len(category_data)):

        previous_row = category_data.iloc[i - 1]
        current_row = category_data.iloc[i]

        growth = current_row["MoM_Growth"]

        if pd.notna(growth) and growth > 20:

            x_values = [
                previous_row["Month"],
                current_row["Month"]
            ]

            y_values = [
                previous_row["Installs"],
                current_row["Installs"]
            ]

            fig.add_trace(
                go.Scatter(
                    x=x_values,
                    y=y_values,
                    mode="lines",
                    line=dict(width=0),
                    fill="tozeroy",
                    fillcolor="rgba(255, 165, 0, 0.20)",
                    name="Growth > 20%",
                    showlegend=False,
                    hoverinfo="skip"
                )
            )

# ------------------------------------------------------------
# 11. Chart formatting
# ------------------------------------------------------------

fig.update_layout(
    title="Task 3: Monthly Install Trends by App Category",
    xaxis_title="Month",
    yaxis_title="Total Installs",
    template="plotly_white",
    hovermode="x unified",
    legend_title="App Category"
)

# ------------------------------------------------------------
# 12. Display only between 6 PM and 9 PM IST
# ------------------------------------------------------------

current_time = datetime.now(
    ZoneInfo("Asia/Kolkata")
).time()

start_time = current_time.replace(
    hour=18,
    minute=0,
    second=0,
    microsecond=0
)

end_time = current_time.replace(
    hour=21,
    minute=0,
    second=0,
    microsecond=0
)

print("\nCurrent IST time:", current_time.strftime("%H:%M:%S"))

if start_time <= current_time < end_time:

    fig.show()

else:

    print(
        "\nTask 3 Time-Series Chart is available only "
        "between 6:00 PM and 9:00 PM IST."
    )
```

```python
# ============================================================
# TASK 5 — DATA PREPARATION
# ============================================================

import pandas as pd
import numpy as np

# 1. Load original Play Store dataset
file_path = "Play Store Data.csv.xlsx"
task5_df = pd.read_excel(file_path)

print("Task 5 dataset loaded successfully!")
print("Rows:", len(task5_df))
print("Columns:")
print(task5_df.columns.tolist())


# ------------------------------------------------------------
# 2. Convert required columns
# ------------------------------------------------------------

task5_df["Rating"] = pd.to_numeric(
    task5_df["Rating"], errors="coerce"
)

task5_df["Reviews"] = pd.to_numeric(
    task5_df["Reviews"], errors="coerce"
)

# Clean Installs column if it contains + or ,
task5_df["Installs"] = (
    task5_df["Installs"]
    .astype(str)
    .str.replace("+", "", regex=False)
    .str.replace(",", "", regex=False)
)

task5_df["Installs"] = pd.to_numeric(
    task5_df["Installs"], errors="coerce"
)


# ------------------------------------------------------------
# 3. Convert Size into Size_MB
# ------------------------------------------------------------

def convert_size_to_mb(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if value == "Varies with device":
        return np.nan

    try:
        if value.endswith("M"):
            return float(value[:-1])

        elif value.endswith("k"):
            return float(value[:-1]) / 1024

        else:
            return np.nan

    except:
        return np.nan


task5_df["Size_MB"] = task5_df["Size"].apply(
    convert_size_to_mb
)


# ------------------------------------------------------------
# 4. Convert Last Updated to datetime
# ------------------------------------------------------------

task5_df["Last Updated"] = pd.to_datetime(
    task5_df["Last Updated"],
    errors="coerce"
)


# ------------------------------------------------------------
# 5. Apply Task 5 filters
# ------------------------------------------------------------

task5_filtered = task5_df[
    (task5_df["Rating"] >= 4.0) &
    (task5_df["Size_MB"] >= 10) &
    (task5_df["Last Updated"].dt.month == 1)
].copy()


# ------------------------------------------------------------
# 6. Remove missing required values
# ------------------------------------------------------------

task5_filtered = task5_filtered.dropna(
    subset=[
        "App",
        "Category",
        "Rating",
        "Reviews",
        "Installs",
        "Size_MB",
        "Last Updated"
    ]
)


# ------------------------------------------------------------
# 7. Aggregate by category
# ------------------------------------------------------------

task5_category = (
    task5_filtered
    .groupby("Category", as_index=False)
    .agg(
        Average_Rating=("Rating", "mean"),
        Total_Reviews=("Reviews", "sum"),
        Total_Installs=("Installs", "sum")
    )
)


# ------------------------------------------------------------
# 8. Select Top 10 categories by installs
# ------------------------------------------------------------

task5_category = (
    task5_category
    .sort_values(
        "Total_Installs",
        ascending=False
    )
    .head(10)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 9. Display results
# ------------------------------------------------------------

print("\nTask 5 data prepared successfully!")
print("=" * 55)

print("\nRows after Task 5 filters:")
print(len(task5_filtered))

print("\nTop 10 categories by total installs:")
print(task5_category)

print("\nMissing values:")
print(
    task5_category[
        [
            "Category",
            "Average_Rating",
            "Total_Reviews",
            "Total_Installs"
        ]
    ].isna().sum()
)
```

```python
# ============================================================
# TASK 6 — DATA PREPARATION
# Use the dataset already loaded for Task 5
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Copy the already-loaded original dataset
# ------------------------------------------------------------

task6_df = task5_df.copy()

print("Task 6 dataset loaded successfully!")
print("Rows:", len(task6_df))
print("Columns:")
print(task6_df.columns.tolist())


# ------------------------------------------------------------
# 2. Convert required numeric columns
# ------------------------------------------------------------

task6_df["Rating"] = pd.to_numeric(
    task6_df["Rating"],
    errors="coerce"
)

task6_df["Reviews"] = pd.to_numeric(
    task6_df["Reviews"],
    errors="coerce"
)


# ------------------------------------------------------------
# 3. Clean Installs
# ------------------------------------------------------------

task6_df["Installs"] = (
    task6_df["Installs"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)

task6_df["Installs"] = pd.to_numeric(
    task6_df["Installs"],
    errors="coerce"
)


# ------------------------------------------------------------
# 4. Clean Price
# ------------------------------------------------------------

task6_df["Price"] = (
    task6_df["Price"]
    .astype(str)
    .str.replace("$", "", regex=False)
)

task6_df["Price"] = pd.to_numeric(
    task6_df["Price"],
    errors="coerce"
)


# ------------------------------------------------------------
# 5. Convert Size to MB
# ------------------------------------------------------------

def convert_size_to_mb(value):

    value = str(value).strip()

    if value.lower() in ["nan", "null", "none"]:
        return np.nan

    if value.endswith("M"):
        return float(value[:-1])

    if value.endswith("k"):
        return float(value[:-1]) / 1024

    return np.nan


task6_df["Size_MB"] = task6_df["Size"].apply(
    convert_size_to_mb
)


# ------------------------------------------------------------
# 6. Extract Android version
# ------------------------------------------------------------

task6_df["Android_Version"] = (
    task6_df["Android Ver"]
    .astype(str)
    .str.extract(r"(\d+(?:\.\d+)?)")[0]
)

task6_df["Android_Version"] = pd.to_numeric(
    task6_df["Android_Version"],
    errors="coerce"
)


# ------------------------------------------------------------
# 7. Calculate Revenue
# ------------------------------------------------------------

task6_df["Revenue"] = (
    task6_df["Installs"] *
    task6_df["Price"]
)


# ------------------------------------------------------------
# 8. Apply Task 6 filters
# ------------------------------------------------------------

task6_filtered = task6_df[
    (task6_df["Installs"] >= 10000) &
    (task6_df["Revenue"] >= 10000) &
    (task6_df["Android_Version"] > 4.0) &
    (task6_df["Size_MB"] > 15) &
    (task6_df["Content Rating"] == "Everyone") &
    (task6_df["App"].astype(str).str.len() <= 30)
].copy()


# ------------------------------------------------------------
# 9. Top 3 categories by total installs
# ------------------------------------------------------------

top3_categories = (
    task6_filtered
    .groupby("Category")["Installs"]
    .sum()
    .sort_values(ascending=False)
    .head(3)
    .index
    .tolist()
)


task6_filtered = task6_filtered[
    task6_filtered["Category"].isin(top3_categories)
].copy()


# ------------------------------------------------------------
# 10. Free / Paid
# ------------------------------------------------------------

task6_filtered["App_Type"] = np.where(
    task6_filtered["Type"].astype(str).str.upper() == "PAID",
    "Paid",
    "Free"
)


# ------------------------------------------------------------
# 11. Summary data
# ------------------------------------------------------------

task6_summary = (
    task6_filtered
    .groupby(["Category", "App_Type"])
    .agg(
        Average_Installs=("Installs", "mean"),
        Total_Revenue=("Revenue", "sum")
    )
    .reset_index()
)


# ------------------------------------------------------------
# 12. Display results
# ------------------------------------------------------------

print("\n=======================================================")
print("Task 6 data prepared successfully!")
print("=======================================================")

print("\nRows after Task 6 filters:")
print(len(task6_filtered))

print("\nTop 3 categories:")
print(top3_categories)

print("\nFree vs Paid:")
print(task6_filtered["App_Type"].value_counts())

print("\nTask 6 summary:")
print(task6_summary)

print("\nMissing values:")
print(task6_summary.isna().sum())
```
